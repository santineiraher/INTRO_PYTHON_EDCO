## 3. Funciones

### Ejercicio 10 (guiado): Función con docstring y anotaciones
Escribe `def area_triangulo(base: float, altura: float) -> float:` que devuelva el área.


````python
# ✅ Solución propuesta
def area_triangulo(base: float, altura: float) -> float:
    """Calcula el área de un triángulo.

    Fórmula: 0.5 * base * altura
    """
    return 0.5 * base * altura

area_triangulo(8, 5)
````

### Ejercicio 11 (guiado): Valores por defecto y `*args`, `**kwargs`
Crea `def potencia(x, y=2)` que eleva `x` a `y`. Demuestra cómo capturar argumentos variables.


````python
# ✅ Solución propuesta

def potencia(x, y=2):
    return x ** y

print(potencia(3))      # por defecto al cuadrado
print(potencia(2, 10))  # potencia alta

# *args y **kwargs demo

def demo_args(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

demo_args(1, 2, 3, nombre="Santiago", curso="Intro Datos")
````

### Ejercicio 12 (guiado): Función que usa otras funciones
Usa `area_triangulo` dentro de `def area_total(figuras)` que recibe una lista de pares `(base, altura)` y devuelve la suma de áreas.


````python
# ✅ Solución propuesta

def area_total(figuras):
    total = 0
    for base, altura in figuras:
        total += area_triangulo(base, altura)
    return total

area_total([(3, 4), (10, 2.5), (8, 8)])
````

### Ejercicio 13 (guiado): Lambda, `map`, `filter`
- Usa `map` para convertir una lista de temperaturas °C a °F.
- Usa `filter` para quedarte con las temperaturas °F mayores a 80.


````python
# ✅ Solución propuesta

c = [18, 20, 22, 25, 28, 30]
a_f = list(map(lambda t: (t * 9/5) + 32, c))
calientes = list(filter(lambda f: f > 80, a_f))
a_f, calientes
````

## 4. Mini‑proyecto (guiado): Resumen de calificaciones
Tenemos una lista de diccionarios con estudiantes y notas.  
Objetivo: calcular promedio, estado (Aprobado/Reprobado), y el mejor estudiante.


````python
# ✅ Solución propuesta

datos = [
    {"nombre": "Ana", "notas": [4.5, 3.9, 4.2]},
    {"nombre": "Luis", "notas": [3.2, 3.5, 2.9]},
    {"nombre": "María", "notas": [4.8, 4.7, 4.9]},
]

def promedio(lista):
    return sum(lista) / len(lista) if lista else 0.0

resumen = []
for d in datos:
    prom = promedio(d["notas"])
    estado = "Aprobado" if prom >= 3.0 else "Reprobado"
    resumen.append({"nombre": d["nombre"], "promedio": round(prom, 2), "estado": estado})

print("Resumen:")
for r in resumen:
    print(r)

mejor = max(resumen, key=lambda x: x["promedio"])
print("\nMejor estudiante:", mejor)
````
