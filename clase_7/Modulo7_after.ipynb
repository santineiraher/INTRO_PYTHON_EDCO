### Regresión Lineal
La regresión lineal es una técnica estadística utilizada para analizar la relación entre una variable una variable dependiente ($Y$) y un conjunto de variables explicativas ($X$). El objetivo de la regresión lineal es encontrar la mejor línea recta que pueda explicar la relación entre las variables y predecir el valor de $Y$ para cualquier valor dado de $X$.

La ecuación de la línea recta se representa como:

$$Y = \alpha + \beta X$$

Donde $\alpha$ es la intersección en $Y$, $\beta$ es la pendiente de la línea recta y $X$ es el valor de la variable independiente. La pendiente de la línea recta indica cuánto aumenta (o disminuye) Y por cada unidad de cambio en $X$.

La regresión lineal se utiliza comúnmente en la investigación y en los negocios para predecir el valor de una variable dependiente en función de una variable independiente. Por ejemplo, se puede utilizar la regresión lineal para predecir la cantidad de ventas que una empresa puede esperar para un determinado nivel de publicidad.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
propiedades = pd.read_csv('./data/co_properties.csv.zip', compression='zip')
propiedades.head()
```

### Manipulación inicial
Suponga que usted, como experto en precio de vivienda en bogotá quiere entender qué determina los precios en general para un predio. Se va a enfocar en predios que sean Apartamentos y Casas, que están en la Zona Noroccidental de la ciudad, que son de tipo Venta y que están entre los 100 y 1'000 millones de pesos. 

```python
## Limpieza de datos - manipulación inicial.
propiedades = propiedades[propiedades.currency=='COP']
propiedades = propiedades.dropna(subset='surface_covered')
propiedades = propiedades[(propiedades.l3=='Bogotá D.C') &(propiedades.l4=='Zona Noroccidental')]
propiedades = propiedades[(propiedades.price<1e9)&(propiedades.surface_covered<500)&(propiedades.price>1e8)]
propiedades = propiedades[(propiedades.property_type.isin(['Apartamento','Casa']))]
propiedades = propiedades[propiedades.operation_type=='Venta']
```

```python
propiedades.shape
```

```python
sns.jointplot(data=propiedades,x='surface_covered',y='price');
```

```python
#!pip install statsmodels
import statsmodels.api as sm
```

```python
y = propiedades['price']
X = propiedades['surface_covered']
```

$$Y = \beta X$$

```python
X = sm.add_constant(X)
```

$$Y = \alpha + \beta X$$

```python
X
```

```python
#OLS -> Regresión (Ordinary Least Squares - Mínimos Cuadrados Ordinarios)
modelo1 = sm.OLS(endog = y, exog = X)
```

```python
# Entrenamos
resultado1 = modelo1.fit()
```

```python
type(resultado1)
```

```python
resultado1.summary()
```

$$precio = \alpha + \beta*surface\_covered$$

```python
resultado1.params/1e6
```

```python
alpha = resultado1.params['const']/1e6
beta_surface_covered = resultado1.params['surface_covered']/1e6
print(f'alpha (millones): ${alpha:.2f}, beta (millones): ${beta_surface_covered:.2f}')
```

```python
superficie = 88
precio = float(alpha+beta_surface_covered*superficie)
precio
```

```python
# Construyamos el ajuste dentro de muestra
y_hat = resultado1.predict()
```

```python
y_hat
```

```python
X
```

```python
prediccion = X.copy()
prediccion['y_pred'] = y_hat
prediccion['y_real'] = y
prediccion
```

```python
import seaborn as sns
```

```python
# Gráfico automático
fig, ax = plt.subplots(figsize = (10, 6))
sns.regplot(x = "surface_covered", y = "price", data = propiedades, ax = ax,
            scatter_kws = {"color": "darkblue", "s": 3}, 
            line_kws = {"color": "red", "linestyle": "--", "linewidth": 1})
plt.legend(["Real", "Predicho", "Intervalo de Confianza"])
plt.xlabel("Superficie ($m^2$)")
plt.ylabel("Precio (COP)")
```

## Clustering

Hacer clustering implica agrupar datos similares en grupos o "clusters" basados en sus características. La idea es permitir que los datos se organicen de manera natural sin etiquetas predefinidas, facilitando el análisis y la interpretación de patrones dentro de los datos. Vamos a ver un ejercicio simple de segmentación. 


```python
# Una agrupación sencilla con KMeans

# 1. Importar librerías necesarias
from sklearn.cluster import KMeans
```

```python
# 2. Seleccionar variables numéricas relevantes para el clustering
# Usaremos superficie cubierta y precio
X_cluster = propiedades[['surface_covered', 'price']]
```

```python
plt.scatter(X_cluster['surface_covered'], X_cluster['price']/1e6, s=10)
plt.xlabel('Superficie cubierta (m²)')
plt.ylabel('Precio (COP - Millones)')
plt.title('Dispersión de propiedades sin estandarización')
```

```python
# 3. Estandarizar las variables (opcional pero recomendado)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)
```

¿Qué es estandarizar? Estandarizar es el proceso de transformar datos para que tengan una media de cero y una desviación estándar de uno. Esto se logra restando la media de cada valor y dividiendo por la desviación estándar. Estandarizar es importante en análisis de datos y aprendizaje automático porque asegura que todas las características contribuyan de manera equitativa al modelo, evitando que las variables con escalas más grandes dominen el análisis.


$$x_{estándar} = \frac{x - \mu}{\sigma}$$


donde  $\mu$ es la media y $\sigma$ es la desviación estándar.

```python
X_scaled
```

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], s=10)
plt.xlabel('Superficie cubierta (estandarizada)')
plt.ylabel('Precio (estandarizado)')
plt.title('Dispersión de propiedades con estandarización')
plt.show()
```

```python
# 4. Elegir el número de clusters (por ejemplo, 3)
kmeans = KMeans(n_clusters=3, random_state=42)
```

```python
propiedades['cluster'] = kmeans.fit_predict(X_scaled)
```

```python
# 5. Visualizar los clusters
plt.figure(figsize=(10,6))
for i in range(3):
    plt.scatter(
        propiedades[propiedades['cluster']==i]['surface_covered'],
        propiedades[propiedades['cluster']==i]['price'],
        label=f'Cluster {i}'
    )
plt.xlabel('Superficie cubierta (m2)')
plt.ylabel('Precio (COP)')
plt.title('Clustering de propiedades por superficie y precio')
plt.legend()
plt.show()
```

### Post processing


```python
# Estadísticas descriptivas por cluster
resumen_clusters = propiedades.groupby('cluster')[['surface_covered', 'price']].agg(
    ['mean', 'median', 'std', 'min', 'max']
).round(2)

resumen_clusters
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))

sns.boxplot(data=propiedades, x='cluster', y='surface_covered', ax=ax[0])
ax[0].set_title('Distribución de superficie por cluster')

sns.boxplot(data=propiedades, x='cluster', y='price', ax=ax[1])
ax[1].set_title('Distribución de precio por cluster')

plt.show()
```
